```python
### !st step crating a pipeline 

## separate data into multiple data categorires 
```

```python
Numeric_data=['Age','Balance','Day','Duration','Campaign','Pdays','Previous']

Ordinal_data=['Education']

Nominal_data=['Job','Marital','Housing','Loan','Default','poutcome','contact','Month']
```

```python
### seperate pipeline transformer for pipline data

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder,OrdinalEncoder
from sklearn.preprocessing import StandardScaler
```

```python
Numeric_pipeline = Pipeline(steps=[('impute',SimpleImputer(strategy='mean')),('scaling',StandardScaler())])
```

```python
Ordinal_pipeline=Pipeline(steps=[('imputer',SimpleImputer(strategy='most_frequent')),('encoding',OrdinalEncoder(categories=[['unknown','primary','secondary','tertiary']])),('scaling',StandardScaler())])
```

```python
Nominal_pipeline=Pipeline(steps=[('imputer',SimpleImputer(strategy='most_frequent')),('encoding',OneHotEncoder(handle_unknown='ignore'))])
```

```python
## 3rd step from herer

## combine these pipeline transformer wiht this repecetive data
#3 import column transformer
```

```python
from sklearn.compose import ColumnTransformer
```

```python
features=ColumnTransformer(transformers=[('Numeric Transformation',Numeric_pipeline,Numeric_data),
('Ordinal Transformation',Ordinal_pipeline,Ordinal_data),
('Nominal Transformation',Nominal_pipeline,Nominal_data)])
```

```python
features
```

```python
### now 4th create model with respective transformer and along with their algorithm
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model=Pipeline(steps=[('features',features),('classifer',LogisticRegression())])
```

```python
model
```
